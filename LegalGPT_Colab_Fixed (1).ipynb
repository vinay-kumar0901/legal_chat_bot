## Start the LegalGPT web application
Run this after Ollama and the FAISS index are ready. This cell starts Uvicorn and checks the health endpoint.

```python
import os
import sys
import time
import subprocess
import requests
from pathlib import Path
from google.colab import output

PROJECT = "/content/legal_chat_bot-main"
os.chdir(PROJECT)

# Ensure required frontend folder exists (the fixed ZIP includes static/index.html).
static_dir = Path(PROJECT) / "static"
static_dir.mkdir(exist_ok=True)
if not (static_dir / "index.html").exists():
    raise FileNotFoundError("static/index.html is missing. Re-upload the fixed LegalGPT ZIP.")

# Stop a previous backend if this cell is rerun.
subprocess.run(["bash", "-lc", "fuser -k 8000/tcp"], capture_output=True)
time.sleep(2)

server_log = open("/content/legalgpt-server.log", "w")
server_process = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "app:app",
     "--host", "0.0.0.0", "--port", "8000"],
    cwd=PROJECT,
    stdout=server_log,
    stderr=subprocess.STDOUT
)

# Wait for FastAPI health endpoint.
for attempt in range(30):
    if server_process.poll() is not None:
        server_log.flush()
        print(open("/content/legalgpt-server.log").read()[-10000:])
        raise RuntimeError("Uvicorn stopped. See the error log above.")
    try:
        r = requests.get("http://127.0.0.1:8000/healthz", timeout=3)
        r.raise_for_status()
        print("LegalGPT backend is healthy:", r.json())
        break
    except requests.RequestException:
        time.sleep(2)
else:
    server_log.flush()
    print(open("/content/legalgpt-server.log").read()[-10000:])
    raise RuntimeError("Backend did not become ready within 60 seconds.")

web_url = output.eval_js("google.colab.kernel.proxyPort(8000)")
print("Open your LegalGPT application:", web_url)
```
